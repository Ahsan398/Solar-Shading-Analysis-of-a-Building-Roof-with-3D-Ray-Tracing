# Tree Data Extraction and Analysis

This notebook demonstrates how to load, inspect, spatially filter, and extract tree positions and heights from the local GeoPackage (`.gpkg`) tree database (`Trees/125015_baeume.gpkg`).

## Dataset Overview
The tree database contains approximately **3.79 million trees** with the following attributes:
- `baumhoehe`: Height of the tree in meters.
- `dgmhoehe`: Digital Terrain Model (DGM) ground height at the tree position in meters.
- `geometry`: Point geometries representing tree locations in coordinate system **EPSG:25832** (ETRS89 / UTM zone 32N).

### 1. Import Libraries

```python
import os
import pandas as pd
import geopandas as gpd
import contextily as cx
import matplotlib.pyplot as plt
from shapely.geometry import box
from pyproj import Transformer
```

### 2. Set Paths and Configuration

```python
TREE_GPKG_PATH = '../Trees/125015_baeume.gpkg'
OUTPUT_DIR = 'processing_output/'
os.makedirs(OUTPUT_DIR, exist_ok=True)
```

### 3. Load the Tree Dataset
Since the full dataset contains nearly 3.8 million points, we load it using `geopandas.read_file`.

```python
print("Loading tree dataset (this may take 10-15 seconds)...")
trees_gdf = gpd.read_file(TREE_GPKG_PATH)
print(f"Loaded {len(trees_gdf):,} trees.")
print("CRS of the dataset:", trees_gdf.crs)
trees_gdf.head()
```

### 4. Spatially Filter Trees for the Study Area
To avoid working with all 3.8 million trees, we define a query box enclosing the Würzburg study area (`X: 561000` to `575000` and `Y: 5505000` to `5523000`) and filter the trees using spatial indexing.

```python
# Define the study area bounding box (UTM Zone 32N)
study_area = box(561000, 5505000, 575000, 5523000)

# Perform spatial filter
print("Filtering trees for the Würzburg study area...")
wuerzburg_trees = trees_gdf[trees_gdf.geometry.intersects(study_area)].copy()
print(f"Found {len(wuerzburg_trees):,} trees in the study area.")
wuerzburg_trees.head()
```

### 5. Extract Coordinate Information
We extract explicit X/Y coordinates in UTM (meters) and transform them to Latitude/Longitude (WGS84 degrees) for easier mapping or use in solar simulation models (e.g. PVlib).

```python
# Extract UTM coordinates
wuerzburg_trees['x_utm'] = wuerzburg_trees.geometry.x
wuerzburg_trees['y_utm'] = wuerzburg_trees.geometry.y

# Transform to Lat/Lon (EPSG:4326)
transformer = Transformer.from_crs("EPSG:25832", "EPSG:4326", always_xy=True)
wuerzburg_trees['longitude'], wuerzburg_trees['latitude'] = transformer.transform(
    wuerzburg_trees['x_utm'].values,
    wuerzburg_trees['y_utm'].values
)

wuerzburg_trees.head()
```

### 6. Inspect Tree Heights and Statistics

```python
print("Tree Height (baumhoehe) statistics (meters):")
print(wuerzburg_trees['baumhoehe'].describe())

print("\nGround Elevation (dgmhoehe) statistics (meters):")
print(wuerzburg_trees['dgmhoehe'].describe())
```

### 7. Export Extracted Tree Data
We save the extracted dataset as a CSV file and a GeoJSON file inside the output directory.

```python
csv_out_path = os.path.join(OUTPUT_DIR, 'wuerzburg_extracted_trees.csv')
geojson_out_path = os.path.join(OUTPUT_DIR, 'wuerzburg_extracted_trees.geojson')

# Export as CSV
wuerzburg_trees.drop(columns=['geometry']).to_csv(csv_out_path, index=False)
print(f"Exported CSV to {csv_out_path}")

# Export as GeoJSON
wuerzburg_trees.to_file(geojson_out_path, driver='GeoJSON')
print(f"Exported GeoJSON to {geojson_out_path}")
```

### 8. Visualize a Subset of Extracted Trees
We plot a spatial distribution map of a small subset of the filtered trees (e.g. 5,000 trees) with a basemap.

```python
# Select a random subset to keep the plot responsive
subset = wuerzburg_trees.sample(n=min(5000, len(wuerzburg_trees)), random_state=42)

fig, ax = plt.subplots(figsize=(10, 10))
sc = ax.scatter(subset['x_utm'], subset['y_utm'], c=subset['baumhoehe'], cmap='viridis', s=10, alpha=0.8)
plt.colorbar(sc, label='Tree Height (meters)')
ax.set_title('Spatial Distribution and Heights of Extracted Trees in Würzburg (Sample of 5,000)')
ax.set_xlabel('UTM Easting (m)')
ax.set_ylabel('UTM Northing (m)')

# Add a background map using contextily
try:
    cx.add_basemap(ax=ax, crs=wuerzburg_trees.crs)
except Exception as e:
    print("Could not add basemap:", e)

plt.show()
```
