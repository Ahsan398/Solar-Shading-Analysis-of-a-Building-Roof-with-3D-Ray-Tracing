# Solar Ray-Tracing with pvlib and PyVista

This notebook performs an annual solar ray-tracing simulation for a building's roof to determine the impact of shading from surrounding buildings and trees throughout a full year.

## Objectives
1. **Load Spatial Data**: Read surrounding building geometries (from CityGML JSON) and tree positions (from a CSV file).
2. **Construct 3D Scene**: Build a single PyVista 3D scene containing surrounding buildings and tree cones.
3. **Compute Annual Sun Positions**: Use `pvlib` to calculate the sun azimuth and elevation for the site at 10-minute intervals for an entire year.
4. **Perform Ray-Tracing**: Cast a ray from each roof panel point toward the sun for each time step. If the ray intersects any building or tree cone, the point is shaded; otherwise, it is sunlit.
5. **Summarize and Visualize**: Output shading statistics for each roof point and visualize the shading patterns in 3D.

```python
import os
import json
import numpy as np
import pandas as pd
import pyvista as pv
import pvlib
from earcut import earcut

print("All libraries imported successfully!")
```

## 1. Simulation Setup and Parameters

In this section, we define the key simulation constants:
- **Site Location**: Latitude, longitude, and timezone of Würzburg, Germany.
- **Target Building ID**: The identifier of the building whose roof we want to analyze (`DEBY_LOD2_506720`).
- **Obstacle Search Radius**: The radius (in meters) around the target building to search for shading obstacles (trees and other buildings). We use `100` meters.
- **Time Horizon**: A full year (2026) sampled at 10-minute intervals.

```python
# Site configuration (Würzburg, Germany)
LATITUDE = 49.77243
LONGITUDE = 9.87073
TIMEZONE = 'Europe/Berlin'

# Target building of interest
TARGET_BUILDING_ID = 'DEBY_LOD2_506720'

# Search radius for obstacles (meters)
OBSTACLE_RADIUS_M = 100.0

# Define input file paths
if os.path.basename(os.getcwd()) == "notebooks":
    DATA_DIR = "../outputs"
else:
    DATA_DIR = "outputs"

TREES_CSV = os.path.join(DATA_DIR, "wuerzburg_extracted_trees.csv")
ROOFS_CSV = os.path.join(DATA_DIR, "wuerzburg_roof_details.csv")
SURFACES_JSON = os.path.join(DATA_DIR, "cityGML_obj_surfaces.json")
OBJECTS_CSV = os.path.join(DATA_DIR, "cityGML_objects.csv")

print(f"Data directory: {DATA_DIR}")
print(f"Trees CSV: {TREES_CSV}")
print(f"Roofs CSV: {ROOFS_CSV}")
```

## 2. Helper Functions for Geometry and Triangulation

To construct 3D meshes of buildings, we need to parse CityGML coordinate strings and triangulate 3D polygons. We define the following helper functions:
- `parse_coords_string`: Parses a space-separated string of 3D coordinates (`"x y z x y z ..."`) into a `(N, 3)` numpy array.
- `find_best_orthogonal_projection_plane`: Fits a 2D plane to 3D coordinates using Singular Value Decomposition (SVD) to find the plane of least variance (e.g. projection onto XY, XZ, or YZ).
- `triangulate_polygon_3d`: Projects a 3D polygon onto its best fitting 2D plane and triangulates it using the `earcut` algorithm.
- `building_to_mesh`: Reconstructs a full 3D building mesh from its Ground, Wall, and Roof surfaces.
- `sample_roof_points`: Extracts roof polygons and samples points (the main centroid and triangle centroids) on the roof surfaces to use as ray-tracing origins.

```python
def parse_coords_string(s: str) -> np.ndarray:
    """Parses a coordinate string 'x y z x y z ...' into an (N,3) array."""
    vals = list(map(float, s.split()))
    pts = np.array(vals, dtype=float).reshape(-1, 3)
    # If the polygon is closed (first point equals last), drop the last duplicate point
    if len(pts) > 1 and np.allclose(pts[0], pts[-1]):
        pts = pts[:-1]
    return pts

def find_best_orthogonal_projection_plane(points: np.ndarray):
    """Finds the best 2D projection plane (XY, XZ, or YZ) for a set of 3D points."""
    # Perform SVD on the centered points
    U, s, Vh = np.linalg.svd(points, full_matrices=False)
    least_principal_axis = Vh[-1, :]
    
    # The coordinate axis most aligned with the least variance is the one to remove
    axis_to_remove_index = np.argmax(np.abs(least_principal_axis))
    
    if axis_to_remove_index == 0:
        return [1, 2], "YZ-plane"
    elif axis_to_remove_index == 1:
        return [0, 2], "XZ-plane"
    else:
        return [0, 1], "XY-plane"

def triangulate_polygon_3d(vertices: np.ndarray) -> np.ndarray:
    """Triangulates a 3D planar polygon by projecting to 2D and using earcut."""
    if len(vertices) < 3:
        return np.empty((0, 3), dtype=int)
    
    # Center points to compute best projection plane
    centered = vertices - np.mean(vertices, axis=0)
    indices_to_keep, _ = find_best_orthogonal_projection_plane(centered)
    
    projected = vertices[:, indices_to_keep]
    flat_coords = projected.flatten().tolist()
    
    # Triangulate using earcut
    faces_idx = earcut.earcut(flat_coords)
    return np.array(faces_idx, dtype=int).reshape(-1, 3)

def building_to_mesh(building_id: str, building_data: dict) -> pv.PolyData:
    """Builds a single PyVista PolyData mesh for a building."""
    all_vertices = []
    all_faces = []
    vertex_offset = 0
    
    # Accumulate ground, wall, and roof surfaces
    for surface_type in ['CoordsGroundSurfaces', 'CoordsWallSurfaces', 'CoordsRoofSurfaces']:
        if surface_type not in building_data:
            continue
        for surf_str in building_data[surface_type]:
            v = parse_coords_string(surf_str)
            if len(v) < 3:
                continue
            faces = triangulate_polygon_3d(v)
            if len(faces) == 0:
                continue
            
            all_vertices.append(v)
            all_faces.append(faces + vertex_offset)
            vertex_offset += len(v)
            
    if len(all_vertices) == 0:
        return None
    
    vertices = np.vstack(all_vertices)
    faces = np.vstack(all_faces)
    
    # Convert face indices to PyVista format: [n_points, i0, i1, i2, ...]
    n_faces = faces.shape[0]
    faces_pv = np.empty((n_faces, 4), dtype=int)
    faces_pv[:, 0] = 3
    faces_pv[:, 1:] = faces
    faces_pv = faces_pv.ravel()
    
    return pv.PolyData(vertices, faces_pv)

def sample_roof_points(roof_coords_str: str) -> np.ndarray:
    """Samples points on the roof surface (centroid and triangle centroids)."""
    vertices = parse_coords_string(roof_coords_str)
    
    # 1. Centroid of the polygon
    centroid = vertices.mean(axis=0)
    sampled = [centroid]
    
    # 2. Centroids of individual triangles forming the polygon
    faces = triangulate_polygon_3d(vertices)
    for face in faces:
        tri_centroid = vertices[face].mean(axis=0)
        sampled.append(tri_centroid)
        
    # Keep only unique points
    unique_points = []
    for p in sampled:
        if not any(np.allclose(p, up) for up in unique_points):
            unique_points.append(p)
            
    return np.array(unique_points)

print("Helper functions defined successfully!")
```

## 3. Loading Spatial Data and Filtering

We load the building data and tree details. To keep the simulation efficient, we compute the centroid of our target building and select only buildings and trees that fall within our obstacle search radius (100 meters).

```python
print("Loading CSV files...")
df_objects = pd.read_csv(OBJECTS_CSV)
df_trees = pd.read_csv(TREES_CSV)
df_roofs = pd.read_csv(ROOFS_CSV)

# Extract coordinates and target location information
target_building = df_objects[df_objects['CityGML_ID'] == TARGET_BUILDING_ID]
if target_building.empty:
    raise ValueError(f"Target building {TARGET_BUILDING_ID} not found in objects dataset!")

target_row = target_building.iloc[0]
cx = target_row['Centroid x UTM']
cy = target_row['Centroid y UTM']
lat = target_row['Centroid Latitude']
lon = target_row['Centroid Longitude']
print(f"Target building centroid UTM coordinates: X={cx:.2f}, Y={cy:.2f}")
print(f"Target location: Latitude={lat:.5f}, Longitude={lon:.5f}")

# Filter surrounding buildings within search radius
dists_b = np.sqrt((df_objects['Centroid x UTM'] - cx)**2 + (df_objects['Centroid y UTM'] - cy)**2)
nearby_buildings = df_objects[dists_b <= OBSTACLE_RADIUS_M]
nearby_b_ids = nearby_buildings['CityGML_ID'].tolist()
print(f"Found {len(nearby_b_ids)} buildings within {OBSTACLE_RADIUS_M}m radius.")

# Filter trees within search radius
dists_t = np.sqrt((df_trees['Tree_Position_X'] - cx)**2 + (df_trees['Tree_Position_Y'] - cy)**2)
df_nearby_trees = df_trees[dists_t <= OBSTACLE_RADIUS_M]
print(f"Found {len(df_nearby_trees)} trees within {OBSTACLE_RADIUS_M}m radius.")
```

## 4. Reconstructing 3D Obstacle Scene Mesh

We now read the raw CityGML surface coordinates for the nearby buildings and build their PyVista meshes. We also construct vertical cone meshes for each tree inside our study area.

According to requirements:
- The cone must model the tree's spread equally in all horizontal directions (360°), i.e., a circular base with radius `cone_radius`.
- Center the cone at `(Tree_Position_X, Tree_Position_Y, dgmhoehe + cone_height/2)`.
- Use `height = cone_height` and `radius = cone_radius`.
- Set `direction = (0, 0, 1)` (vertical direction).

Finally, all meshes are merged into a single `PolyData` mesh `obstacle_mesh` using `pv.merge` for high performance ray tracing.

```python
print("Loading building surface details (JSON)...")
with open(SURFACES_JSON, 'r') as f:
    surfaces_data = json.load(f)

# Reconstruct building meshes
building_meshes = []
for b_id in nearby_b_ids:
    if b_id in surfaces_data:
        mesh = building_to_mesh(b_id, surfaces_data[b_id])
        if mesh is not None:
            building_meshes.append(mesh)
print(f"Constructed {len(building_meshes)} building meshes.")

# Reconstruct tree cone meshes
tree_meshes = []
for idx, row in df_nearby_trees.iterrows():
    h = row['cone_height']
    r = row['cone_radius']
    tx = row['Tree_Position_X']
    ty = row['Tree_Position_Y']
    tz_ground = row['dgmhoehe']
    
    # Position center of the cone at (X, Y, Ground_Z + Height / 2)
    center = (tx, ty, tz_ground + h / 2.0)
    
    # Build vertical cone mesh
    cone = pv.Cone(
        center=center, 
        direction=(0, 0, 1), 
        height=h, 
        radius=r, 
        resolution=16
    )
    tree_meshes.append(cone)
print(f"Constructed {len(tree_meshes)} tree cone meshes.")

# Merge all meshes into a single PyVista obstacle mesh
print("Merging all meshes into a single PyVista PolyData obstacle mesh...")
all_meshes = building_meshes + tree_meshes
if len(all_meshes) == 0:
    raise ValueError("No meshes to combine in the scene!")

obstacle_mesh = pv.merge(all_meshes)
print(f"Obstacle scene constructed!")
print(f"  Total points: {obstacle_mesh.points.shape[0]}")
print(f"  Total cells: {obstacle_mesh.n_cells}")
```

## 5. Annual Solar Vector Calculations

Using `pvlib.solarposition.get_solarposition`, we calculate the solar angles (azimuth and elevation) for Würzburg for the entire year at 10-minute intervals.
We then convert these angles into 3D unit direction vectors representing the rays pointing from the roof points towards the Sun.

Note:
- Elevation is the angle above the horizon.
- Azimuth is the angle measured clockwise from North (Y-axis).
- Z-component: $\sin(\text{elevation})$
- Y-component: $\cos(\text{elevation}) \cdot \cos(\text{azimuth})$
- X-component: $\cos(\text{elevation}) \cdot \sin(\text{azimuth})$

```python
# Generate timeline for the year 2026 at 10-minute intervals
timeline = pd.date_range(
    start='2026-01-01 00:00:00', 
    end='2026-12-31 23:50:00', 
    freq='10min', 
    tz=TIMEZONE
)
print(f"Total timeline steps: {len(timeline)}")

# Compute solar position using pvlib
print("Computing solar position (this may take a few seconds)...")
solar_pos = pvlib.solarposition.get_solarposition(timeline, lat, lon)

# Select columns and add to dataframe
df_solar = pd.DataFrame(index=timeline)
df_solar['elevation'] = solar_pos['elevation']
df_solar['azimuth'] = solar_pos['azimuth']

# Filter day vs night (Sun above horizon)
df_solar['is_day'] = df_solar['elevation'] > 0.0
daytime_steps = df_solar['is_day'].sum()
print(f"Daytime steps: {daytime_steps} ({daytime_steps / len(timeline) * 100:.1f}%)")

# Pre-calculate unit vectors for daytime steps
elevation_rad = np.radians(df_solar['elevation'])
azimuth_rad = np.radians(df_solar['azimuth'])

df_solar['sun_dx'] = np.cos(elevation_rad) * np.sin(azimuth_rad)
df_solar['sun_dy'] = np.cos(elevation_rad) * np.cos(azimuth_rad)
df_solar['sun_dz'] = np.sin(elevation_rad)

df_solar.head(10)
```

## 6. Roof Points Selection

We extract the target roof details for building `DEBY_LOD2_506720` and sample points across each roof surface. We will simulate shading for each of these points.

```python
target_roofs = df_roofs[df_roofs['Building_ID'] == TARGET_BUILDING_ID]
print(f"Target building {TARGET_BUILDING_ID} has {len(target_roofs)} roof surfaces.")

roof_points = []
roof_labels = []

for _, row in target_roofs.iterrows():
    r_idx = row['Roof_Index']
    coords_str = row['roof_coords']
    pitch = row['pitch']
    orientation = row['orientation']
    
    pts = sample_roof_points(coords_str)
    print(f"Roof Index {r_idx} (Pitch: {pitch:.1f}°, Orientation: {orientation:.1f}°): Sampled {len(pts)} points.")
    
    for i, pt in enumerate(pts):
        roof_points.append(pt)
        roof_labels.append(f"Roof_{r_idx}_Pt_{i}")

roof_points = np.array(roof_points)
print(f"Total roof points to simulate: {len(roof_points)}")
```

## 7. Ray-Tracing Simulation Loop

For each roof point and each time step, we cast a ray from the roof point towards the Sun.
To avoid false self-intersection with the starting roof polygon, we nudge the origin along the Sun vector by `0.05` meters (5 centimeters).
If the ray hits the combined `obstacle_mesh`, the point is shaded; otherwise, it is sunlit.
All night steps are set to shaded automatically without running the ray-tracer to optimize computation.

```python
import time

shading_matrix = np.ones((len(timeline), len(roof_points)), dtype=bool) # Default: all shaded (True)
print("Starting annual ray-tracing simulation...")
start_time = time.time()

# Identify indices where it is day
daytime_indices = np.where(df_solar['is_day'].values)[0]
num_daytime = len(daytime_indices)
print(f"Simulating {num_daytime} daytime steps for {len(roof_points)} roof points...")

# Extract pre-calculated sun vectors
sun_vectors = df_solar[['sun_dx', 'sun_dy', 'sun_dz']].values

# We run a loop over daytime steps
for count, step_idx in enumerate(daytime_indices):
    sun_vec = sun_vectors[step_idx]
    
    for pt_idx, pt in enumerate(roof_points):
        # Nudge the origin along the solar ray to avoid starting-surface intersection
        origin = pt + sun_vec * 0.05
        end_point = origin + sun_vec * 1000.0  # Cast ray out to 1km
        
        # Ray trace query
        pts, cells = obstacle_mesh.ray_trace(origin, end_point)
        
        # If no intersection, the point is sunlit (False for shading)
        if len(pts) == 0:
            shading_matrix[step_idx, pt_idx] = False
            
    # Progress indicator
    if (count + 1) % 5000 == 0 or (count + 1) == num_daytime:
        elapsed = time.time() - start_time
        print(f"  Processed {count + 1}/{num_daytime} steps ({ (count + 1)/num_daytime * 100:.1f}%). Elapsed: {elapsed:.1f}s")

end_time = time.time()
print(f"Ray-tracing simulation completed in {end_time - start_time:.1f} seconds!")
```

## 8. Summary of Shading Results

Let's compile the results and display the annual shading statistics. We'll show the percentage of time steps that each point is sunlit vs. shaded.

```python
# Convert results to a pandas DataFrame
df_results = pd.DataFrame(index=timeline)
for pt_idx, label in enumerate(roof_labels):
    df_results[label] = shading_matrix[:, pt_idx]

# Calculate statistics for each roof point
summary_data = []
for label in roof_labels:
    shaded_steps = df_results[label].sum()
    total_steps = len(df_results)
    sunlit_steps = total_steps - shaded_steps
    
    # Calculate daytime stats specifically
    daytime_shaded = df_results.loc[df_solar['is_day'], label].sum()
    daytime_total = df_solar['is_day'].sum()
    daytime_sunlit = daytime_total - daytime_shaded
    
    summary_data.append({
        'Point Label': label,
        'Annual Sunlit Hours': sunlit_steps * 10 / 60.0,
        'Annual Shaded Hours': shaded_steps * 10 / 60.0,
        'Sunlit % (Overall)': sunlit_steps / total_steps * 100.0,
        'Sunlit % (Daylight Hours)': daytime_sunlit / daytime_total * 100.0
    })

df_summary = pd.DataFrame(summary_data)
print("Annual Solar Shading Summary:")
display(df_summary.round(2))
```

## 9. 3D Visualization of Shading Patterns

We visualize the target building, surrounding buildings, tree cones, and the sampled roof points. The points are colored by their annual sunlit percentage during daylight hours to show shading patterns.

```python
# Create a PyVista plotter
# Set pv.OFF_SCREEN = True to make it work in headless environments
pv.OFF_SCREEN = True

plotter = pv.Plotter(window_size=(1024, 768))

# Add the combined obstacle mesh (buildings and trees) with custom styling
plotter.add_mesh(
    obstacle_mesh, 
    color='lightgray', 
    opacity=0.85, 
    show_edges=True, 
    edge_color='dimgray',
    label='Surroundings (Obstacles)'
)

# Extract sunlit percentages for coloring
sunlit_percentages = df_summary['Sunlit % (Daylight Hours)'].values

# Create roof points mesh
points_mesh = pv.PolyData(roof_points)
points_mesh['Sunlit %'] = sunlit_percentages

# Add roof points with a color map representing sunlit percentage
plotter.add_mesh(
    points_mesh,
    scalars='Sunlit %',
    cmap='plasma',
    point_size=15.0,
    render_points_as_spheres=True,
    clim=[0, 100],
    scalar_bar_args={'title': 'Sunlit % (Daylight Hours)'}
)

# Focus camera on the target building
plotter.camera_position = [
    (cx - 50, cy - 80, 380),  # Camera position
    (cx, cy, 335),            # Focus point (target centroid)
    (0, 0, 1)                 # View up vector
]

# Set background and light
plotter.set_background('#334155')
plotter.add_light(pv.Light(position=(cx - 100, cy - 100, 500), intensity=0.8))

print("Plotter initialized and scene built!")
# Save screenshot of the scene
screenshot_path = os.path.join(DATA_DIR, "shading_visualization.png")
plotter.screenshot(screenshot_path)
print(f"Screenshot saved to {screenshot_path}")
```
